### OBJECTIVE

This program estimates the sample size required, for instance when conducting a market survey or poll.
The confidence interval and margin of error are required, as wel as an estimate of the expected proportion of responses, a default of 50$ is provided, though any proportion below 50% can be used.

The program uses a GUI created from tkinter for inputting the above metrics. It was originally written using the Sublime text editor and has run on Jupyter notebooks. Iin its present form it does not run in GitHub or Google Colab.

```python
import tkinter as tk
import math
from tkinter import *
from tkinter import messagebox
from tkinter import ttk

"""
========================================================================
Set up GUI and get inputs
========================================================================
"""

root = tk.Tk()
root.title("Simple Stat GUI")
root.geometry("300x200")

padx_constant = 20
pady_constant = 20

# Create an overall frame
frame = tk.Frame(root)
frame.pack()


# Creaate  the first user input frame - User_info_frame_1

class SampleSize_Class:

    def __init__(self, root):

        # ------------------------------------------------ Initialize variables -----------------------------------------

        self.Frame_font_tuple = ("Arial", 12, "bold")         # Reference: See https://www.geeksforgeeks.org/how-to-set-font-for-text-in-tkinter/

        # ----------------------------------------------- Dataframe 0 -----------------------------------------------------
        # Dataframe 3 - SAmple Size Message

        self.user_info_frame_S0 = tk.LabelFrame(frame,  text="Sample Size Calculation Overview",  font=self.Frame_font_tuple)
        self.user_info_frame_S0.grid(row=0, column=0, sticky="news", padx=padx_constant, pady=pady_constant)         # 'Sticky' - see Hala video around 30:00

        self.SampleSizeMsg_0 = "Sample Size, given margin of error and confidence level."                           # it required about six tabs  for this line to align with the one below
        self.msg_label3_0 = tk.Label(self.user_info_frame_S0, text = self.SampleSizeMsg_0)
        self.msg_label3_0.grid(row=0, column=0, sticky=tk.W)                                                        # Justify to the West, i.e. Left Justify

        self.SampleSizeMsg_1 = "Assume maximum universe population size when proportion = 50%."
        self.msg_label3_1 = tk.Label(self.user_info_frame_S0, text = self.SampleSizeMsg_1)
        self.msg_label3_1.grid(row=1, column=0, sticky=tk.W)


        # ----------------------------------------------- Initialize Scenario Variables -----------------------------------------------------
        # Initialize input variable labels
        self.First_Param = "Margin of Error"
        self.Second_Param = "Confidence Level"
        self.Third_Param = "Best Proportion"

        self.Result_Msg = "Sample SizeX: "

        print(f"\nIn main..")

        # Create and initialize input data variables

        # Scenario 1
        self.in_dat1_1 = DoubleVar()                        # first input data field is float
        self.in_dat2_1 = DoubleVar()                        # second input data field is float
        self.in_dat3_1 = IntVar()                           # third input data field is integer

        self.in_dat1_1.set(2.0)
        self.in_dat2_1.set(95)
        self.in_dat3_1.set(50)

        # Scenario 2
        self.in_dat1_2 = DoubleVar()                        # first input data field is float
        self.in_dat2_2 = DoubleVar()                        # second input data field is float
        self.in_dat3_2 = IntVar()                           # third input data field is integer

        self.in_dat1_2.set(2.0)
        self.in_dat2_2.set(95)
        self.in_dat3_2.set(50)

        # Scenario 3
        self.in_dat1_3 = DoubleVar()                        # first input data field is float
        self.in_dat2_3 = DoubleVar()                        # second input data field is float
        self.in_dat3_3 = IntVar()                           # third input data field is integer

        self.in_dat1_3.set(2.0)
        self.in_dat2_3.set(95)
        self.in_dat3_3.set(50)


        self.Calc_Result = DoubleVar()                      # to hold the result of the calculations
        self.Res_S1 = DoubleVar()                           # to hold the result of the calculations
        self.Res_S2 = DoubleVar()                           # to hold the result of the calculations
        self.Res_S3 = DoubleVar()                           # to hold the result of the calculations


        self.submit_button = tk.Button(root, text="SUBMIT", command=self.run_submit)
        self.submit_button.pack(pady=20)                 # double check what 'pady' does'

        # -------------------------------------------------- Overall Run Structure ---------------------------------------------------------

    def run_SampleSize(self):
        print(f"\nIn run method.")

        self.mycounter = 0

        # Scenario 1
        self.user_info_frame = tk.LabelFrame(frame,  text="Sample Size Estimate: Scenario 1", font=self.Frame_font_tuple)
        self.user_info_frame.grid(row=1, column=0, padx=padx_constant, pady=pady_constant)
        self.SampleSizePrep(self.in_dat1_1, self.in_dat2_1, self.in_dat3_1)
        self.SampleSizeCalc(self.in_dat1_1, self.in_dat2_1, self.in_dat3_1)
        print(f"\n ~~~ In run_SampleSize, Scenario {self.mycounter}:  MargOfError_ is {self.MargOfError_},  TargetProp_  is {self.TargetProp_}, Confidence level is {self.Conf_level}.")
        print(f"\n ~~~ In run_SampleSize:  Sample size result for Scenario {self.mycounter} is {self.Calc_Result}.")

        # Scenario 2
        self.user_info_frame = tk.LabelFrame(frame,  text="Sample Size Estimate: Scenario 2", font=self.Frame_font_tuple)
        self.user_info_frame.grid(row=2, column=0, padx=padx_constant, pady=pady_constant)
        self.SampleSizePrep(self.in_dat1_2, self.in_dat2_2, self.in_dat3_2)
        self.SampleSizeCalc(self.in_dat1_2, self.in_dat2_2, self.in_dat3_2)
        print(f"\n ~~~ In run_SampleSize, Scenario {self.mycounter}:  MargOfError_ is {self.MargOfError_},  TargetProp_  is {self.TargetProp_}, Confidence level is {self.Conf_level}.")
        print(f"\n ~~~ In run_SampleSize:  Sample size result for Scenario {self.mycounter} is {self.Calc_Result}.")

        # Scenario 3
        self.user_info_frame = tk.LabelFrame(frame,  text="Sample Size Estimate: Scenario 3", font=self.Frame_font_tuple)
        self.user_info_frame.grid(row=3, column=0, padx=padx_constant, pady=pady_constant)
        self.SampleSizePrep(self.in_dat1_3, self.in_dat2_3, self.in_dat3_3)
        self.SampleSizeCalc(self.in_dat1_3, self.in_dat2_3, self.in_dat3_3)
        print(f"\n ~~~ In run_SampleSize, Scenario {self.mycounter}:  MargOfError_ is {self.MargOfError_},  TargetProp_  is {self.TargetProp_}, Confidence level is {self.Conf_level}.")
        print(f"\n ~~~ In run_SampleSize:  Sample size result for Scenario {self.mycounter} is {self.Calc_Result}.")


    # Prepare entry and message widgets for input variables
    def SampleSizePrep(self, in_dat1, in_dat2, in_dat3):
        print (f"\n\nP-P-P In SampleSizePrep and my counter is {self.mycounter}.")
        self.msg_label = tk.Label(self.user_info_frame, text = self.First_Param)
        self.msg_label.grid(row=0, column=0)

        self.msg_label = tk.Label(self.user_info_frame, text = self.Second_Param)
        self.msg_label.grid(row=0, column=1)

        self.msg_label = tk.Label(self.user_info_frame, text = self.Third_Param)
        self.msg_label.grid(row=0, column=2)

        # create an output to store the results
        self.result_message_1 = tk.Message(self.user_info_frame, width=100, text=self.Result_Msg, fg="blue")
        self.result_message_1.grid(row=1, column=3)


        print(f"\n In SampleSizePrep: In_dat1 is {self.in_dat1_1} and In_dat2 is {self.in_dat2_1}.")

        self.Enter1 = tk.Entry(self.user_info_frame, width=20, textvariable=in_dat1)
        self.Enter1.grid(row=1, column=0)

        self.Enter2 = ttk.Combobox(self.user_info_frame, value=['80', '85', '90','95','99','99.9'], width=20, textvariable=in_dat2)
        # self.Enter2.set(95)
        self.Enter2.grid(row=1, column=1)

        self.Enter3 = tk.Spinbox(self.user_info_frame, from_=10, to=50, width=20, textvariable=in_dat3)
        self.Enter3.grid(row=1, column=2)

        # space out all fields using user_info_frame
        for widget in self.user_info_frame.winfo_children():
            widget.grid_configure(padx=10, pady=5)


    # --------------------------------------------- Sample Siae Calc -------------------------------------------

    def SampleSizeCalc(self, in_dat1, in_dat2, in_dat3):
        print(f"\n\nC-C-C In SampleSizeCalc method and My counter is {self.mycounter}.")


        self.MargOfError = in_dat1.get()                # 2.10
        self.Conf_level  = in_dat2.get()                # 90, 95, 95
        self.TargetProp  = in_dat3.get()                # 50

        self.MargOfError_ = self.MargOfError/ 100
        self.TargetProp_ = self.TargetProp/ 100
        self.n_SampleSize_ = 0
        self.ConfLevelFlag = False


        self.Conf_level_Dct     = {80:1.28, 85:1.44, 90:1.645, 95:1.960, 99:2.576, 99.9:3.291}                    # Confidence level dictionary

        print(f"\nIn SampleSizeCalc method, after variable declaration, before calculation.")
        print(f"\nIn SAmpleSizeCalc: In_dat1 (Margin of error) is {self.in_dat1_1} and In_dat2 (Confidence level )is {self.in_dat2_1}.")
        print(f"\nMargin of error is {self.MargOfError} and Cofnince level is {self.Conf_level}.")

        for key, value in self.Conf_level_Dct.items():                    # Get the z-Score equivalent to the confidence level inputted
            if  self.Conf_level ==  key:
                self.z_Score = value
                self.ConfLevelFlag = True
                print(f"\nCofnidence level is {self.Conf_level} and z_Score is  {self.z_Score}.")
            else:
                # do nothing, leave the Confidence level flag false
                pass


        if self.ConfLevelFlag == True:
            self.Calc_Result = round((self.z_Score**2 * (self.TargetProp_ * (1 - self.TargetProp_))) / (self.MargOfError_)**2)
            self.result_message_1.config(text=f"Sample Size: {self.Calc_Result}")
            print(f"\n +++ In scenario {self.mycounter} smargin of error is {self.MargOfError_}.")


    def run_submit(self):
        print (f"\n\n>>> Executing run_submit method.")
        self.run_SampleSize()
        print(f"\n --- My counter is {self.mycounter}, Calc_Result is {self.Calc_Result}.")


if __name__ == "__main__":
    ExecSampleSize = SampleSize_Class(root)
    ExecSampleSize.run_SampleSize()
    root.mainloop()

```
